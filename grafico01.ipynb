```python
import matplotlib.pyplot as plt
import numpy as np

# Generamos 30,000 puntos aleatorios
x = np.random.rand(30000)
y = np.random.rand(30000)

# Creamos el gráfico de dispersión
plt.figure(figsize=(10, 6))
plt.scatter(x, y, s=1, alpha=0.5, color='blue')  # s=1 para puntos pequeños

# Títulos y etiquetas
plt.title("Gráfico de dispersión con 30,000 puntos")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")

# Mostrar gráfico
plt.show()
```
